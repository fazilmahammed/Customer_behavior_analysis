```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','-')
df=df.rename(columns={'purchase-amount-(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels=['Young Adult','Adult','Middle-aged','Senior']
df['age_grouped']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age_grouped','age']].head(10)
```

```python
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency-of-purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency-of-purchases']].head(10)
```
